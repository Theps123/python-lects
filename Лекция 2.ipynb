```python
def monkey_count(n):
    return list(range(1, n + 1))
```

```python
def positive_sum(arr):
    return sum(x for x in arr if x > 0)
```

```python
def basic_op(operator, value1, value2):
    if operator == '+':
        return value1 + value2
    elif operator == '-':
        return value1 - value2
    elif operator == '*':
        return value1 * value2
    elif operator == '/':
        return value1 / value2
```
